```python
# Python basics
print("Hello, Data Mining!")
```

```python
# Variables and data types
student_age = 21
cgpa = 3.75
student_name = "Amit Bhowmik"
is_student = True

print("Student Name:", student_name)
print("Age:", student_age)
print("CGPA:", cgpa)
print("Is Student:", is_student)

print("\nData Types:")
print(type(student_age))
print(type(cgpa))
print(type(student_name))
print(type(is_student))
```

```python
# Operators
a = 20
b = 6

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Power:", a ** b)

print("\nComparison Operators:")
print("a > b:", a > b)
print("a < b:", a < b)
print("a == b:", a == b)
print("a != b:", a != b)
```

```python
# Lists
courses = ["Data Mining", "Database", "Artificial Intelligence", "Computer Networks"]
print("Courses:", courses)
print("First Course:", courses[0])
print("Last Course:", courses[-1])

courses.append("Machine Learning")
print("After Adding:", courses)

courses.remove("Database")
print("After Removing:", courses)
```

```python
# Tuples
student_info = ("Amit Bhowmik", "CSE", 21, 3.75)

print("Student Information:", student_info)
print("Name:", student_info[0])
print("Department:", student_info[1])
print("Age:", student_info[2])
print("CGPA:", student_info[3])
```

```python
# Dictionaries
student = {
    "name": "Amit Bhowmik",
    "department": "CSE",
    "age": 21,
    "cgpa": 3.75
}

print(student)
print("Name:", student["name"])
print("Department:", student["department"])
print("CGPA:", student["cgpa"])
```

```python
# Conditional statements
cgpa = 3.75

if cgpa >= 3.50:
    print("Excellent academic performance")
elif cgpa >= 3.00:
    print("Good academic performance")
elif cgpa >= 2.50:
    print("Average academic performance")
else:
    print("Needs improvement")
```

```python
# For loop
for i in range(1, 11):
    print(i)
```

```python
# Loop through a list
subjects = ["Data Mining", "Operating System", "Computer Networks", "Database"]

for subject in subjects:
    print("Subject:", subject)
```

```python
# While loop
count = 1

while count <= 5:
    print("Count:", count)
    count += 1
```

```python
# Functions
def calculate_average(mark1, mark2, mark3):
    return (mark1 + mark2 + mark3) / 3

result = calculate_average(80, 85, 90)
print("Average Mark:", result)
```

```python
# NumPy basics
import numpy as np

marks = np.array([75, 80, 85, 90, 95])

print("Marks:", marks)
print("Mean:", np.mean(marks))
print("Maximum:", np.max(marks))
print("Minimum:", np.min(marks))
print("Standard Deviation:", np.std(marks))
```

```python
# Pandas DataFrame
import pandas as pd

student_data = {
    "Name": ["Amit", "Rahim", "Karim", "Hasan"],
    "Age": [21, 22, 20, 23],
    "CGPA": [3.75, 3.40, 3.90, 3.20]
}

student_df = pd.DataFrame(student_data)
student_df
```
