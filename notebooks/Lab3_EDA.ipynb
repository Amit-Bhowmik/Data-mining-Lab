### Correlation Matrix

```python
# Select numerical columns

numeric_df = df.select_dtypes(include=np.number).copy()

# Remove Student_ID because it is only an identifier
if "Student_ID" in numeric_df.columns:
    numeric_df = numeric_df.drop(columns=["Student_ID"])

print("Numerical Features:")
print(numeric_df.columns.tolist())
```

```python
# Calculate correlation matrix

correlation_matrix = numeric_df.corr()

correlation_matrix
```

```python
# Correlation Heatmap

plt.figure(figsize=(12, 9))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Features")

plt.show()
```

### Interpretation

**The correlation coefficient ranges from -1 to +1:**

- +1 → strong positive relationship  
- -1 → strong negative relationship  
- 0 → little or no linear relationship  

For example, GPA, Semester GPA, and CGPA may show strong relationships because they represent related academic performance measures.

### Outlier Detection
We will use the Interquartile Range (IQR) method.

```python
# Detect GPA outliers using IQR

Q1 = df["GPA"].quantile(0.25)
Q3 = df["GPA"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["GPA"] < lower_bound) |
    (df["GPA"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of GPA Outliers:", len(outliers))
```

```python
# Detect outliers in important numerical features

features = [
    "Age",
    "Study_Hours_per_Day",
    "Attendance_Rate",
    "Assignment_Delay_Days",
    "Travel_Time_Minutes",
    "Stress_Index",
    "GPA",
    "Semester_GPA",
    "CGPA"
]

print("Outlier Analysis")
print("=" * 50)

for column in features:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ).sum()

    print(f"{column}: {outlier_count} outliers")
```

### Analyze Relationships

```python
# Find strongest relationships between numerical features

corr = numeric_df.corr()

correlation_pairs = corr.unstack()

# Remove self-correlations
correlation_pairs = correlation_pairs[
    correlation_pairs.index.get_level_values(0)
    != correlation_pairs.index.get_level_values(1)
]

# Sort by absolute correlation
correlation_pairs = correlation_pairs.abs().sort_values(
    ascending=False
)

print("Top 10 Strongest Relationships:")
print(correlation_pairs.head(10))
```

### Dropout Analysis
This section is especially useful because your project's main goal is student dropout prediction.

```python
# Display dropout counts

print("Dropout Distribution:")
print(df["Dropout"].value_counts())
```

```python
# Calculate dropout percentages

dropout_percentage = (
    df["Dropout"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("Dropout Percentage:")
print(dropout_percentage)
```

```python
# Average GPA according to dropout status

average_gpa = df.groupby("Dropout")["GPA"].mean()

print("Average GPA by Dropout Status:")
print(average_gpa)
```

```python
# Average Attendance according to dropout status

average_attendance = df.groupby("Dropout")["Attendance_Rate"].mean()

print("Average Attendance by Dropout Status:")
print(average_attendance)
```

```python
# Average Study Hours according to dropout status

average_study_hours = df.groupby("Dropout")["Study_Hours_per_Day"].mean()

print("Average Study Hours by Dropout Status:")
print(average_study_hours)
```
