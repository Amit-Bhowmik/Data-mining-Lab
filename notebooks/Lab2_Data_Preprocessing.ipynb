```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder, MinMaxScaler, StandardScaler

import warnings
warnings.filterwarnings("ignore")
```

```python
path = "data/student_dropout.csv"
df = pd.read_csv(path)

print("Dataset loaded successfully!")
```

```python
df.head()
```

```python
df.tail()
```

```python
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
```

```python
print("Dataset Columns:\n")
for i, column in enumerate(df.columns, start=1):
    print(i, "-", column)
```

```python
df.dtypes
```

```python
df.info()
```

```python
missing_values = df.isnull().sum()
print(missing_values)
print("\nTotal Missing Values:", df.isnull().sum().sum())
```

```python
# Handle missing values
missing_before = df.isnull().sum().sum()
print("Missing values before handling:", missing_before)

numeric_columns = df.select_dtypes(include=np.number).columns
for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

categorical_columns = df.select_dtypes(exclude=np.number).columns
for column in categorical_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].mode()[0])

missing_after = df.isnull().sum().sum()
print("Missing values after handling:", missing_after)
```

```python
duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)
```

```python
df = df.drop_duplicates()
print("Dataset shape after removing duplicates:", df.shape)
```

```python
df.describe()
```

```python
categorical_columns = df.select_dtypes(exclude=np.number).columns
print("Categorical Columns:")
print(categorical_columns.tolist())
```

```python
# Encode the categorical target
print("Dropout values before encoding:")
print(df["Dropout"].unique())
```

```python
label_encoder = LabelEncoder()

df["Dropout_Encoded"] = label_encoder.fit_transform(df["Dropout"])

print("Dropout values after encoding:")
print(df[["Dropout", "Dropout_Encoded"]].head())
```

```python
print("Encoding Mapping:")
for number, category in enumerate(label_encoder.classes_):
    print(category, "=", number)
```

```python
# Normalization
numeric_columns = df.select_dtypes(include=np.number).columns

minmax_scaler = MinMaxScaler()
normalized_data = minmax_scaler.fit_transform(df[numeric_columns])

normalized_df = pd.DataFrame(normalized_data, columns=numeric_columns)
normalized_df.head()
```

```python
print("Minimum values:")
print(normalized_df.min().head())

print("\nMaximum values:")
print(normalized_df.max().head())
```

```python
# Standardization
standard_scaler = StandardScaler()
standardized_data = standard_scaler.fit_transform(df[numeric_columns])

standardized_df = pd.DataFrame(standardized_data, columns=numeric_columns)
standardized_df.head()
```

```python
print("Mean of standardized data:")
print(standardized_df.mean().head())

print("\nStandard deviation:")
print(standardized_df.std().head())
```

```python
output_path = "data/processed_dataset.csv"
df.to_csv(output_path, index=False)
print("Processed dataset saved successfully!")
```

## Conclusion

The dataset was loaded, examined, cleaned, encoded, scaled, and summarized. Missing values and duplicate records were investigated, the target variable was encoded, and both Min-Max normalization and standardization were demonstrated. The processed dataset was saved for future data mining tasks.
